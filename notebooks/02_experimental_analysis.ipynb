```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

results = pd.read_csv(
    "../results/raw/experiment_results.csv"
)

summary = pd.read_csv(
    "../results/summary/experiment_summary.csv"
)

sns.set_theme(style="whitegrid")

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=summary,
    x="dataset_size",
    y="mean_distance",
    hue="algorithm",
    marker="o"
)

plt.title("Mean TSP Route Distance by Algorithm")
plt.xlabel("Number of Cities")
plt.ylabel("Mean Route Distance")
plt.xticks([10, 20, 30, 40, 50])
plt.legend(title="Algorithm")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 6))

sns.lineplot(
    data=summary,
    x="dataset_size",
    y="mean_execution_time",
    hue="algorithm",
    marker="o"
)

plt.title("Execution Time by Algorithm")
plt.xlabel("Number of Cities")
plt.ylabel("Mean Execution Time (seconds)")
plt.xticks([10, 20, 30, 40, 50])
plt.legend(title="Algorithm")
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(12, 6))

sns.boxplot(
    data=results,
    x="dataset_size",
    y="best_distance",
    hue="algorithm"
)

plt.title("Distribution of Route Distances")
plt.xlabel("Number of Cities")
plt.ylabel("Best Route Distance")
plt.legend(title="Algorithm")
plt.tight_layout()
plt.show()
```
